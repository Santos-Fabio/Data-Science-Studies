### Fixing Values
Cleaning wrong data, aggrupping values, formating values

```python
for i in df.columns:
    print(f'{i}: {df[i].unique()}')
```

#### Removing spaces and formating

```python
# I've done exemples that were 'str' and not 'object', must see first which type it is
for i in df.columns:
    if(df[i].dtype=='object'):
        df[i]= (df[i].str.replace(r'\s+',' ',regex=True).str.strip().str.lower())
```

```python
df.head(2)
```

#### Fixing Numeric Columns
- Age
- AI Usage (h/day)
- Monthly AI Cost
- Timed Saved (h/week)
- Work/Study (h/day)

```python
# Age
df['Age'] = df['Age'].str.replace('yrs','').str.strip()
df['Age'] = pd.to_numeric(df['Age'],downcast='integer',errors='coerce').astype('Int64')
# AI usage
df['AI_Usage_Hours_Per_Day'] = df['AI_Usage_Hours_Per_Day'].str.replace('hrs','').str.strip()
df['AI_Usage_Hours_Per_Day'] = pd.to_numeric(df['AI_Usage_Hours_Per_Day'],errors='coerce')
# Monthly AI Cost
df['Monthly_AI_Cost'] = df['Monthly_AI_Cost'].str.replace('$','').str.strip()
df['Monthly_AI_Cost'] = pd.to_numeric(df['Monthly_AI_Cost'],errors='coerce')
# Timed Saved Per Week
df['Time_Saved_Hours_Per_Week'] = df['Time_Saved_Hours_Per_Week'].str.replace('hrs','').str.strip()
df['Time_Saved_Hours_Per_Week'] = pd.to_numeric(df['Time_Saved_Hours_Per_Week'],errors='coerce')
# Work/Study Hours per Day
df['Work_or_Study_Hours_Per_Day'] = df['Work_or_Study_Hours_Per_Day'].str.replace('hrs','').str.strip()
df['Work_or_Study_Hours_Per_Day'] = pd.to_numeric(df['Work_or_Study_Hours_Per_Day'],errors='coerce')
```

#### FIxing Non-Numeric Columns
- Gender
- Country
- User type
- Education Level
- Profession
- AI Tool
- Would Recommend

```python
# Gender
df['Gender'] = df['Gender'].replace({
    'm': 'male',
    'f': 'female'
})
# Country
df['Country'] = df['Country'].replace({
    'u.k': 'united kingdom',
    'united states of america': 'u.s.a',
    'united states': 'u.s.a'
})
# Education_Level
df['Education_Level'] = df['Education_Level'].replace({
    'master\'s': 'master',
    'master\'s degree': 'master',
    'ma/msc': 'master',
    'bachelor\'s': 'bachelor',
    'bachelor\'s degree': 'bachelor',
    'ba/bsc': 'bachelor',
    'hs': 'high school',
    'ph.d': 'phd',
    'undergrad': 'undergraduate'
})
# AI Tool
df['AI_Tool'] = df['AI_Tool'].replace({
    'chat gpt': 'chatgpt',
    'google gemini': 'gemini',
    'microsoft copilot': 'copilot'
})
```
